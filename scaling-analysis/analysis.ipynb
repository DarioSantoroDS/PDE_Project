```python
import pandas as pd

# 2D no ref and 2D harder

# Function to read and process the CSV data
def load_and_process_csv(file_path):
    # Read the CSV into a DataFrame
    df = pd.read_csv(file_path)
    
    # Drop the 'Warnings' column only if it exists
    if 'Warnings' in df.columns:
        df = df.drop(columns=['Warnings'])
    
    # Return the processed DataFrame
    return df
```

```python
import pandas as pd
from pathlib import Path

def read_study_csv(path):
    """
    Read a study CSV file (study 1, 2, or 3) into a pandas DataFrame.

    Parameters
    ----------
    path : str or pathlib.Path
        Path to the CSV file.

    Returns
    -------
    pandas.DataFrame
        DataFrame containing the study data.
    """
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"CSV file not found: {path}")

    df = pd.read_csv(path)

    # Ensure consistent column order if CSVs are identical (they should be)
    expected_cols = [
        "processors",
        "elements",
        "dofs",
        "fgmres_iterations",
        "total_wallclock_s",
        "assemble_preconditioners_s",
        "assembly_system_s",
        "make_grid_s",
        "setup_dofs_s",
        "solve_iterative_s",
        "real_s",
        "user_s",
        "sys_s",
    ]

    missing = set(expected_cols) - set(df.columns)
    if missing:
        raise ValueError(f"Missing expected columns: {missing}")

    return df[expected_cols]
```

#### Speedup

```python
import matplotlib.pyplot as plt

# Benchmark for 1 process (using row with Number of Processes = 1)
benchmark_row = df_2d_no_ref[df_2d_no_ref['Number of Processes'] == 1].iloc[0]

# Extract benchmark values (all times for 1 process)
benchmark_wallclock_time = benchmark_row['Wallclock Time (s)']
benchmark_solver_time = benchmark_row['Solver Time (s)']
benchmark_assembly_time = benchmark_row['Assembly Time (s)']
benchmark_grid_time = benchmark_row['Grid Time (s)']
benchmark_setup_dofs_time = benchmark_row['Setup DOFs Time (s)']
benchmark_fgmres_iterations = benchmark_row['FGMRES Iterations']

# Calculate speed-up for each row
df_2d_no_ref['Speed-up Wallclock'] = benchmark_wallclock_time / df_2d_no_ref['Wallclock Time (s)']
df_2d_no_ref['Speed-up Solver'] = benchmark_solver_time / df_2d_no_ref['Solver Time (s)']
df_2d_no_ref['Speed-up Assembly'] = benchmark_assembly_time / df_2d_no_ref['Assembly Time (s)']
df_2d_no_ref['Speed-up Grid'] = benchmark_grid_time / df_2d_no_ref['Grid Time (s)']
df_2d_no_ref['Speed-up Setup DOFs'] = benchmark_setup_dofs_time / df_2d_no_ref['Setup DOFs Time (s)']
df_2d_no_ref['Speed-up FGMRES'] = benchmark_fgmres_iterations / df_2d_no_ref['FGMRES Iterations']

# Separate plot for Wallclock Time Speed-up
plt.figure(figsize=(8, 6))
plt.plot(df_2d_no_ref['Number of Processes'], df_2d_no_ref['Speed-up Wallclock'], marker='o', label='Wallclock Time Speed-up', color='b')
plt.xlabel('Number of Processes')
plt.ylabel('Speed-up')
plt.title('Wallclock Time Speed-up')
plt.grid(True)
plt.tight_layout()
plt.savefig('./2D_noRef_out/speed_up_wallclock.pdf')
plt.close()

# Separate plot for Solver Time Speed-up
plt.figure(figsize=(8, 6))
plt.plot(df_2d_no_ref['Number of Processes'], df_2d_no_ref['Speed-up Solver'], marker='o', label='Solver Time Speed-up', color='g')
plt.xlabel('Number of Processes')
plt.ylabel('Speed-up')
plt.title('Solver Time Speed-up')
plt.grid(True)
plt.tight_layout()
plt.savefig('./2D_noRef_out/speed_up_solver.pdf')
plt.close()

# Separate plot for Assembly Time Speed-up
plt.figure(figsize=(8, 6))
plt.plot(df_2d_no_ref['Number of Processes'], df_2d_no_ref['Speed-up Assembly'], marker='o', label='Assembly Time Speed-up', color='r')
plt.xlabel('Number of Processes')
plt.ylabel('Speed-up')
plt.title('Assembly Time Speed-up')
plt.grid(True)
plt.tight_layout()
plt.savefig('./2D_noRef_out/speed_up_assembly.pdf')
plt.close()

# Separate plot for Grid Time Speed-up
plt.figure(figsize=(8, 6))
plt.plot(df_2d_no_ref['Number of Processes'], df_2d_no_ref['Speed-up Grid'], marker='o', label='Grid Time Speed-up', color='c')
plt.xlabel('Number of Processes')
plt.ylabel('Speed-up')
plt.title('Grid Time Speed-up')
plt.grid(True)
plt.tight_layout()
plt.savefig('./2D_noRef_out/speed_up_grid.pdf')
plt.close()

# Separate plot for Setup DOFs Time Speed-up
plt.figure(figsize=(8, 6))
plt.plot(df_2d_no_ref['Number of Processes'], df_2d_no_ref['Speed-up Setup DOFs'], marker='o', label='Setup DOFs Time Speed-up', color='m')
plt.xlabel('Number of Processes')
plt.ylabel('Speed-up')
plt.title('Setup DOFs Time Speed-up')
plt.grid(True)
plt.tight_layout()
plt.savefig('./2D_noRef_out/speed_up_setup_dofs.pdf')
plt.close()

# Separate plot for FGMRES Iterations Speed-up
plt.figure(figsize=(8, 6))
plt.plot(df_2d_no_ref['Number of Processes'], df_2d_no_ref['Speed-up FGMRES'], marker='o', label='FGMRES Iterations Speed-up', color='y')
plt.xlabel('Number of Processes')
plt.ylabel('Speed-up')
plt.title('FGMRES Iterations Speed-up')
plt.grid(True)
plt.tight_layout()
plt.savefig('./2D_noRef_out/speed_up_fgmres.pdf')
plt.close()

# Print statement indicating where the plots have been saved
print("Plots saved as PDFs in './2D_noRef_out' directory.")
```

### Speedup

```python
# =========================
# SPEED-UP (3D_noRef_out) — TRUE FIXED SIZE (LATEX-SAFE)
# =========================
import matplotlib.pyplot as plt
import os

# Output directory
os.makedirs('./3D_noRef_out', exist_ok=True)

# ---- FIXED FIGURE SIZE ----
FIGSIZE = (10, 6)   # inches
DPI = 300

# Column aliases
proc_col = 'Processes'
wall_col = 'Wallclock Time (s)'
solve_col = 'Solve Time (s)'
assembly_col = 'Assembly Time (s)'
prec_col = 'Preconditioner Time (s)'
grid_col = 'Grid Setup Time (s)'
dof_col = 'DOF Setup Time (s)'

# ---- Benchmark (p = 1) ----
benchmark_row = df_3d_no_ref[df_3d_no_ref[proc_col] == 1].iloc[0]

benchmark_wall = benchmark_row[wall_col]
benchmark_solve = benchmark_row[solve_col]
benchmark_assembly = benchmark_row[assembly_col]
benchmark_prec = benchmark_row[prec_col]
benchmark_grid = benchmark_row[grid_col]
benchmark_dof = benchmark_row[dof_col]

# ---- Speed-ups ----
df_3d_no_ref['Speed-up Wallclock'] = benchmark_wall / df_3d_no_ref[wall_col]
df_3d_no_ref['Speed-up Solve'] = benchmark_solve / df_3d_no_ref[solve_col]
df_3d_no_ref['Speed-up Assembly'] = benchmark_assembly / df_3d_no_ref[assembly_col]
df_3d_no_ref['Speed-up Preconditioner'] = benchmark_prec / df_3d_no_ref[prec_col]
df_3d_no_ref['Speed-up Grid'] = benchmark_grid / df_3d_no_ref[grid_col]
df_3d_no_ref['Speed-up DOF'] = benchmark_dof / df_3d_no_ref[dof_col]

# ---- Fixed plotting helper ----
def fixed_speedup_plot(x, y, title, filename):
    fig = plt.figure(figsize=FIGSIZE, dpi=DPI)
    ax = fig.add_axes([0.12, 0.15, 0.80, 0.75])  # FIXED MARGINS

    ax.plot(x, y, marker='o')
    ax.set_xlabel('Number of Processes')
    ax.set_ylabel('Speed-up')
    ax.set_title(title)
    ax.grid(True)

    fig.savefig(filename)
    plt.close(fig)

# ---- Plots ----
fixed_speedup_plot(
    df_3d_no_ref[proc_col],
    df_3d_no_ref['Speed-up Wallclock'],
    'Wallclock Speed-up (3D, no refinement)',
    './3D_noRef_out/speed_up_wallclock.pdf'
)

fixed_speedup_plot(
    df_3d_no_ref[proc_col],
    df_3d_no_ref['Speed-up Solve'],
    'Solve Speed-up (3D, no refinement)',
    './3D_noRef_out/speed_up_solve.pdf'
)

fixed_speedup_plot(
    df_3d_no_ref[proc_col],
    df_3d_no_ref['Speed-up Assembly'],
    'Assembly Speed-up (3D, no refinement)',
    './3D_noRef_out/speed_up_assembly.pdf'
)

fixed_speedup_plot(
    df_3d_no_ref[proc_col],
    df_3d_no_ref['Speed-up Preconditioner'],
    'Preconditioner Speed-up (3D, no refinement)',
    './3D_noRef_out/speed_up_preconditioner.pdf'
)

fixed_speedup_plot(
    df_3d_no_ref[proc_col],
    df_3d_no_ref['Speed-up Grid'],
    'Grid Setup Speed-up (3D, no refinement)',
    './3D_noRef_out/speed_up_grid.pdf'
)

fixed_speedup_plot(
    df_3d_no_ref[proc_col],
    df_3d_no_ref['Speed-up DOF'],
    'DOF Setup Speed-up (3D, no refinement)',
    './3D_noRef_out/speed_up_dof_setup.pdf'
)

print("Speed-up plots saved with IDENTICAL size (LaTeX-safe).")
```

#### Cheatup

```python
# =========================
# CHEAT SPEED-UP (3D_harder)
# =========================
# Parameters
BASELINE_PROCS = 4          # p used as baseline
COMM_REMOVAL_FACTOR = 0.7   # fraction kept after removing communication overhead
#   1.0 -> no removal
#   0.9 -> remove 10%
#   0.8 -> remove 20%

import matplotlib.pyplot as plt
import os

os.makedirs('./3D_harder', exist_ok=True)

proc_col = 'Processes'
wall_col = 'Wallclock time'
prec_col = 'Assemble Preconditioners'
assembly_col = 'Assembly System'
grid_col = 'Make Grid'
dof_col = 'Setup DOFs'
solve_col = 'Solve Iterative'

# Baseline row (p = BASELINE_PROCS)
baseline_row = df_3d_no_ref_harder[
    df_3d_no_ref_harder[proc_col] == BASELINE_PROCS
].iloc[0]

# Compute "cheat" single-process reference times
def cheat_reference(value):
    return COMM_REMOVAL_FACTOR * (value / BASELINE_PROCS)

cheat_wall = cheat_reference(baseline_row[wall_col])
cheat_solve = cheat_reference(baseline_row[solve_col])
cheat_assembly = cheat_reference(baseline_row[assembly_col])
cheat_prec = cheat_reference(baseline_row[prec_col])
cheat_grid = cheat_reference(baseline_row[grid_col])
cheat_dof = cheat_reference(baseline_row[dof_col])

# Compute cheat speed-ups
df_3d_no_ref_harder['Cheat Speed-up Wallclock'] = cheat_wall / df_3d_no_ref_harder[wall_col]
df_3d_no_ref_harder['Cheat Speed-up Solve'] = cheat_solve / df_3d_no_ref_harder[solve_col]
df_3d_no_ref_harder['Cheat Speed-up Assembly'] = cheat_assembly / df_3d_no_ref_harder[assembly_col]
df_3d_no_ref_harder['Cheat Speed-up Preconditioner'] = cheat_prec / df_3d_no_ref_harder[prec_col]
df_3d_no_ref_harder['Cheat Speed-up Grid'] = cheat_grid / df_3d_no_ref_harder[grid_col]
df_3d_no_ref_harder['Cheat Speed-up DOFs'] = cheat_dof / df_3d_no_ref_harder[dof_col]

# ---- Plot helper ----
def plot_cheat(x, y, title, filename):
    plt.figure(figsize=(8, 6))
    plt.plot(x, y, marker='o')
    plt.xlabel('Number of Processes')
    plt.ylabel('Cheat Speed-up')
    plt.title(title)
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(filename)
    plt.close()

# ---- Plots ----
plot_cheat(
    df_3d_no_ref_harder[proc_col],
    df_3d_no_ref_harder['Cheat Speed-up Wallclock'],
    'Cheat Wallclock Speed-up (3D harder)',
    './3D_harder/cheat_speedup_wallclock.pdf'
)

plot_cheat(
    df_3d_no_ref_harder[proc_col],
    df_3d_no_ref_harder['Cheat Speed-up Solve'],
    'Cheat Solve Speed-up (3D harder)',
    './3D_harder/cheat_speedup_solve.pdf'
)

plot_cheat(
    df_3d_no_ref_harder[proc_col],
    df_3d_no_ref_harder['Cheat Speed-up Assembly'],
    'Cheat Assembly Speed-up (3D harder)',
    './3D_harder/cheat_speedup_assembly.pdf'
)

plot_cheat(
    df_3d_no_ref_harder[proc_col],
    df_3d_no_ref_harder['Cheat Speed-up Preconditioner'],
    'Cheat Preconditioner Speed-up (3D harder)',
    './3D_harder/cheat_speedup_preconditioner.pdf'
)

plot_cheat(
    df_3d_no_ref_harder[proc_col],
    df_3d_no_ref_harder['Cheat Speed-up Grid'],
    'Cheat Grid Speed-up (3D harder)',
    './3D_harder/cheat_speedup_grid.pdf'
)

plot_cheat(
    df_3d_no_ref_harder[proc_col],
    df_3d_no_ref_harder['Cheat Speed-up DOFs'],
    'Cheat DOF Setup Speed-up (3D harder)',
    './3D_harder/cheat_speedup_dofs.pdf'
)

print(
    f"Cheat speed-up plots saved in './3D_harder' "
    f"(baseline p={BASELINE_PROCS}, comm factor={COMM_REMOVAL_FACTOR})."
)
```
