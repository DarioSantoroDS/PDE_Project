## Functions

```python
import pandas as pd

# 2D no ref and 2D harder

# Function to read and process the CSV data
def load_and_process_csv(file_path):
    # Read the CSV into a DataFrame
    df = pd.read_csv(file_path)
    
    # Drop the 'Warnings' column only if it exists
    if 'Warnings' in df.columns:
        df = df.drop(columns=['Warnings'])
    
    # Return the processed DataFrame
    return df
```

```python
import pandas as pd
import ast

# 2D refine

# Function to read and process the CSV data
def process_refine_csv(file_path):
    # Read the CSV into a DataFrame
    df = pd.read_csv(file_path)
    
    # Create a new empty DataFrame to store processed data
    processed_data = []
    
    # Iterate through each row of the original dataframe
    for _, row in df.iterrows():
        # Get the number of processes and cycles
        num_processes = row['num_processes']
        cycles = row['cycles']
        
        # Convert the columns with list-like strings to actual lists
        mesh_elements = ast.literal_eval(row['mesh_elements'])
        dofs = ast.literal_eval(row['dofs'])
        wallclock_time = ast.literal_eval(row['wallclock_time'])
        solver_time = ast.literal_eval(row['solver_time'])
        fgmres_iterations = row['fgmres_iterations']
        
        # Create new rows where each list item corresponds to a separate row
        for cycle in range(cycles):
            processed_data.append({
                'num_processes': num_processes,
                'cycles': cycle + 1,
                'mesh_elements': mesh_elements[cycle],
                'dofs': dofs[cycle],
                'wallclock_time': wallclock_time[cycle],
                'solver_time': solver_time[cycle],
                'fgmres_iterations': fgmres_iterations
            })
    
    # Create a DataFrame from the processed data
    processed_df = pd.DataFrame(processed_data)
    
    return processed_df
```

```python
import pandas as pd

# 2D weak

# Function to read and process the 2D_noRef_weak.csv data and split into three DataFrames
def process_no_ref_weak_csv(file_path):
    # Read the CSV into a DataFrame
    df = pd.read_csv(file_path)
    
    # Rename the columns for easier reference
    df.columns = ['num_processes', 'mesh_elements', 'wallclock_time', 'user_time', 'sys_time']
    
    # Split the DataFrame into three separate ones based on row indices
    study_1 = df.iloc[0:3].reset_index(drop=True)  # First 3 rows (study 1)
    study_2 = df.iloc[3:6].reset_index(drop=True)  # Next 3 rows (study 2)
    study_3 = df.iloc[6:].reset_index(drop=True)   # Last rows (study 3)
    
    # Return the three study DataFrames
    return study_1, study_2, study_3
```
